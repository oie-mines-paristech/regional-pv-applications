# Objective

The goal of this notebook is to demonstrate how the [Pan-European Climate Database (PECD) v4.2](https://cds.climate.copernicus.eu/datasets/sis-energy-pecd?tab=overview) solar photovoltaic (SPV) outputs can be reproduced using the **c3s-lot2-spv-conversion** library, which internally exploits [regional-pv](https://github.com/ramaroesilva/regional-pv/). To run this script, users must first download surface solar radiation downwards (ssrd) and 2m temperature (t2m) input data from the [CDS](https://cds.climate.copernicus.eu/datasets/sis-energy-pecd?tab=download).

# Code overview

The climate input data is converted into SPV estimates by using the `c3s_lot2_spv_conversion.interface.compute_spv` function. This function requires the following inputs:

- `pv_typology`: string, describing which set of parametrizatiosn to be considered (more details in [documentation](https://confluence.ecmwf.int/pages/viewpage.action?pageId=540941752#ClimateandenergyrelatedvariablesfromthePanEuropeanClimateDatabasederivedfromreanalysisandclimateprojectionsv4.2:Productuserguide(PUG)-UpscalingtoRegionalPVAggregation)). either the name or the ID can be used:

| ID   | Typology name       |
|------|---------------------|
| "60" | rooftop_industrial  |
| "61" | rooftop_residential |
| "62" | utility_fix         |
| "63" | utility_track_1axis |

- `in_ssrd_path`: string, path to .nc file containing surface solar radiation downwards (ssrd). 

- `in_t2m_path`: string, path to .nc file containing 2m temperature (t2m).

- `in_excl_mask_path`: string to .nc file containing a exclusion mask (see section 2.9.2.4.6 and 2.8 in [documentation](https://confluence.ecmwf.int/pages/viewpage.action?pageId=540941752)).

- `DataStream`: string, `HIST` or `PROJ` to indicate if input data comes from ERA5 or a climate projection.

- `print_progress`: boolean, prints index of pixel being computed, allowing users to see if computations are ongoing

## A side-note

To make this notebook self-sufficient, not demanding input data from users to test this fcuntion, a 1-week subset of ERA5 data is included in the same folder as the notebook. This subsetting in turn reduces the number of computations involved in this example.

# The actual code

```python
import sys

import matplotlib.pyplot as plt

# full path to regional-pv-applications\c3s-lot2-spv-conversion
base_path = "..."  # to be completed by user
sys.path.append(base_path)
import c3s_lot2_spv_conversion  # noqa: E402
```

```python
# inputs

# path to ssrd .nc file
in_ssrd_path = "..."  # to be completed by user
# path to t2m .nc file
in_t2m_path = "..."  # to be completed by user

# pv typology to be computed ("60" corresponds to "industrial_rooftops")
pv_typology = "60"
# data stream being considered ("HIST" corresponds to historical, using ERA5)
DataStream = "HIST"
# path to exclusion mask .nc file
in_excl_mask_path = "default"

out, XX, YY = c3s_lot2_spv_conversion.interface.compute_spv(
    pv_typology,
    in_ssrd_path,
    in_t2m_path,
    in_excl_mask_path,
    DataStream,
    print_progress=True,
    to_save=True,
    out_path=r".\out_may2025_v4.nc",
)
```

```python
plt.pcolor(XX, YY, out.mean(axis=0))
plt.xlabel("Longitude")
plt.xlabel("Latitude")
plt.colorbar()
plt.title("Mean SPV capacity factor over first week of May 2025")
```

As seen in the plot above, the input climate was successfully converted to SPV estimates!
